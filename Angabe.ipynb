## Aufgabe 2
Sie haben herausgefunden, dass ein Reihe an Simulationen in den Ordnern TRAJ_00001 bis TRAJ_00019 mit falschen Parametern gestartet wurden. Aus diesem Grund müssen sie die Daten `data.out` aus jedem TRAJ_ ordner löschen um Platz für die neuen zu machen. Ein Kollege hat Ihnen bereits eine Datei zur Verfügung gestellt, die die korrekten Parameter enthält. Allerdings befindet sich der Paramter "rngseed" darin, der in jeder Simulation und damit jedem Ordner einen anderen Wert haben soll. Löschen Sie `data.out` aus jedem TRAJ Ordner und kopieren Sie die Datei `input` in jeden dieser Ordner wobei Sie die Zahl neben "rngseed" durch einen neue zufällig ausgewählte Integer zwischen -99999999 und 99999999 aus einer uniformen Verteilung ersetzen sollen.

```python
import os
import random

with open("Aufgabe02/input.txt", "r") as datei:
    inhalt = datei.read()

for i in range(0, 20):
    file = f"TRAJ_{i:05d}"
    path = os.path.join("Aufgabe02", file)

    data_path = os.path.join(path, "data.out")

    if os.path.isfile(data_path):
        os.remove(data_path)

    rngseed = random.randint(-99999999, 99999999)
    rows = inhalt.splitlines()
    
    for j, row in enumerate(rows):
        if row.startswith("rngseed"):
            rows[j] = f"rngseed {rngseed}"

    new_trj = "\n".join(rows) + "\n"
    input_path = os.path.join(path, "input")

    with open(input_path, "w") as datei:
        datei.write(new_trj)
```

## Aufgabe 3
Nach der erfolgreichen Simulation wollen Sie herausfinden welche der Trajektorien zu ende gelaufen sind. Dafür können Sie nach der Datei `output.err` suchen. In diesem File steht auch drinnen wie viele erfolgreiche Simulationsschritte stattgefunden haben bevor die Simulation abgestürtzt ist. Finden Sie alle abgestürzten Trajektorien und geben Sie aus wie viele Schritte erfolgreich waren als error aus. Um Speicherplatz zu sparen, sollen Sie die sehr großen `wavefunction.tmp` Datein in jedem Ordner löschen.

```python
import os

base_path = "Aufgabe03"

for i in range(0,20):
    file = f"TRAJ_{i:05d}"
    path = os.path.join(base_path, file)
    
    wavefunction_path = os.path.join(path, "wavefunction.tmp")
    if os.path.isfile(wavefunction_path):
        os.remove(wavefunction_path)
        
    err_path = os.path.join(path, "output.err")
    if os.path.isfile(err_path):
        with open(err_path, "r") as err:
            inhalt = err.read()
            print(f"{err_path}: {inhalt}")
```

```python
import os

base_path = "Aufgabe04"
goat_tf = [False, False, True]

os.chdir("Aufgabe04/Door1")

with open("goat.txt", "w") as datei:
    datei.write("leider keine Auto")

os.chdir("..")

for dir in range(2,4):
    path_dir = f"Door{dir}"
    os.mkdir(path_dir)
    os.chdir(path_dir)
    if goat_tf[dir-1] == False:
        with open("goat.txt", "w") as datei:
            datei.write("leider keine Auto")
    elif goat_tf[dir-1] == True:
        with open("car.txt", "w") as datei:
            datei.write("Sie habem ein Auto gewonnen")
    os.chdir("..")
```
